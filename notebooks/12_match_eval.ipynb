```python
import os
import pandas as pd
from notebooks._utils import calculate_accuracy

dataset_root = "/home/xzhao/workspace/GYB_self-ensemble/datasets"
baselines = [
    "baseline_per_prompt.0shots.temp1.0.topp0.95.repeat1", 
    "baseline_per_prompt.0shots", 
    "baseline_per_prompt.5shots"
]
```

```python
def get_baselines(dataet_name):
    if "thinking" in dataet_name:
        return ["baseline_per_prompt.0shots.temp1.0.topp0.95.repeat1"]
    else:
        return [
            "baseline_per_prompt.0shots", 
            # "baseline_per_prompt.5shots"
        ]
```

```python
models = ["llama3.2_1b", "llama3.2_3b", "llama3.1_8b", "qwen3_1.7b", "qwen3_4b", "qwen3_8b", "qwen3_14b"]
# models = ["qwen3_1.7b", "qwen3_4b", "qwen3_8b", "qwen3_14b"]
# dataset_name = "hotpot-debug"
dataset_name = "myriadlama-debug"

for model in models:
    print(f"=== Evaluating model: {model} ===")
    baselines = get_baselines(dataset_name)
    for baseline in baselines:
        print(f"--- Baseline: {baseline} ---")
        eval_path = f"{dataset_root}/{dataset_name}/{model}/{baseline}"
        for file_name in os.listdir(eval_path):
            # if "thinking" in dataset_name and "2paras" not in file_name:
            #     continue
            # elif "thinking" not in dataset_name and "5paras" not in file_name and "baseline" not in file_name:
            #     continue
            
            if file_name.startswith("baseline"):
                df = pd.read_feather(f"{eval_path}/{file_name}")
                calculate_accuracy(df, by_probs=False, is_multichoice=False, birdirect=True, use_generation=True, label="Baseline accuracy")
            elif file_name.startswith("modifyattn.modifyrope.scalescore"):
                df = pd.read_feather(f"{eval_path}/{file_name}")
                calculate_accuracy(df, by_probs=False, is_multichoice=False, birdirect=True, use_generation=True, label="series ensemble accuracy")
            elif file_name.startswith("parallel.avg.avglayer"):
                df = pd.read_feather(f"{eval_path}/{file_name}")
                calculate_accuracy(df, by_probs=False, is_multichoice=False, birdirect=True, use_generation=True, label="parallel ensemble accuracy")
            elif file_name.startswith("parallel.avg.1samples") or file_name.startswith("parallel.avg.5samples"):
                df = pd.read_feather(f"{eval_path}/{file_name}")
                calculate_accuracy(df, by_probs=False, is_multichoice=False, birdirect=True, use_generation=True, label="logits-level ensemble accuracy")
```
