```python
import pandas as pd
from utils import take_until_punct_or_space

pp = pd.read_feather("/home/xzhao/workspace/GYB_self-ensemble/datasets/myriadlama/llama3.2_3b_it/baseline_per_prompt.5shots.feather")
base = pd.read_feather("/home/xzhao/workspace/GYB_self-ensemble/datasets/myriadlama/llama3.2_3b_it/myriadlama.singleparaqapair.5samples.1paras.feather")
```

```python
cnt = 0
for uuid, paraphrase, prompt, generation in zip(base['uuid'], base['paraphrases'], base['prompt'], base['generation_lemmas']):
    paraphrase = paraphrase.tolist()[0]
    sdf = pp[(pp["uuid"] == uuid) & (pp["paraphrase"] == paraphrase)]
    pp_generation = take_until_punct_or_space(sdf['generation_lemmas'].tolist()[0].tolist())
    generation = take_until_punct_or_space(generation.tolist())
    pp_prompt = sdf['prompt'].tolist()[0]
    if pp_generation != generation:
        cnt += 1
        print("---------- UUID ---------- ")
        # print("base prompt:", prompt)
        print("base generation:", generation)
        # print("pp prompt:", pp_prompt)
        print("pp generation:", pp_generation)
        print("-------------------------- ")
    # break
```
