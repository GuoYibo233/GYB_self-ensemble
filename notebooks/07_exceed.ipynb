```python
import pandas as pd
ensemble_filename = "/home/xzhao/workspace/GYB_self-ensemble/datasets/myriadlama/llama3.1_8b/myriadlama.modifyattn.modifyrope.scalescore.singleparaqapair.5samples.5paras.feather"
perprompt_filename = "/home/xzhao/workspace/GYB_self-ensemble/datasets/myriadlama/llama3.1_8b/baseline_per_prompt.5shots.feather"

ensemble_df = pd.read_feather(ensemble_filename)
perprompt_df = pd.read_feather(perprompt_filename)
```

```python
generation_lemmas = ensemble_df['generation_lemmas'].tolist()
answer_lemmas = [[token.tolist() for token in ans.tolist()] for ans in ensemble_df['answer_lemmas'].tolist()]
uuids = ensemble_df['uuid'].tolist()
paraphrases = ensemble_df['paraphrases'].tolist()

for uuid, gen_lem, ans_lem, paras in zip(uuids, generation_lemmas, answer_lemmas, paraphrases):
    print("-----")
    print(f"UUID: {uuid}")
    # print(f"Paraphrase: {paras}")
    ppdf_thisuuid = perprompt_df[perprompt_df['uuid']==uuid]
    pp_gen_lemmas_list = []
    for para in paras: 
        ppdf_thisuuid_thispara = ppdf_thisuuid[ppdf_thisuuid['paraphrase'] == para]
        pp_gen_lemmas = ppdf_thisuuid_thispara['generation_lemmas'].tolist()[0].tolist()
        pp_gen_lemmas_list.append(pp_gen_lemmas)
    print(f"Answer Lemma: {ans_lem}")
    print(f"Generation Lemma: {gen_lem}")
    for idx, pp_gen_lemmas in enumerate(pp_gen_lemmas_list):
        print(f"{idx}-th Per-prompt Generation Lemmas: {pp_gen_lemmas}")
    break
```
