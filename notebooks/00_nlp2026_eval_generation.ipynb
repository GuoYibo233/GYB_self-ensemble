```python
import os
import pandas as pd
from notebooks._utils import calculate_accuracy

data_root = "/home/xzhao/workspace/GYB_self-ensemble/datasets"
dataset_name = "myriadlama-debug"
num_shot = 5


for model_name in ["qwen3_0.6b", "qwen3_1.7b", "qwen3_4b", "qwen3_8b", "qwen3_14b"]:
    if num_shot == 0:
        baseline_filename = f"{data_root}/{dataset_name}/{model_name}/baseline_per_prompt.0shots.feather"
        ensemble_filename = f"{data_root}/{dataset_name}/{model_name}/{dataset_name.split('-')[0]}.logits.avg.0fshots.1samples.-1paras.feather"
    else:
        baseline_filename = f"{data_root}/{dataset_name}/{model_name}/baseline_per_prompt.5shots.feather"
        ensemble_filename = f"{data_root}/{dataset_name}/{model_name}/{dataset_name.split('-')[0]}.logits.avg.1samples.-1paras.feather"
    
    print("========================================")
    print(f"Dataset: {dataset_name.split('-')[0]}, Model: {model_name}, Num_fewshots: {num_shot}")
        
    if os.path.exists(baseline_filename) is False:
        print(f"Baseline file {baseline_filename} does not exist.")
    else:
        baseline_df = pd.read_feather(baseline_filename)
        org_df = baseline_df[baseline_df['is_orig']].reset_index(drop=True)
        
        calculate_accuracy(
            baseline_df, is_multichoice=False,
            birdirect=True, use_generation=True, by_probs=False,
            label="Averaged accuracy")

        calculate_accuracy(
            org_df, is_multichoice=False,
            birdirect=True, use_generation=True, by_probs=False,
            label="Accuracy of raw prompt")

    if os.path.exists(ensemble_filename):
        ensemble_df = pd.read_feather(ensemble_filename)
        calculate_accuracy(
            ensemble_df, is_multichoice=False,
            birdirect=True, use_generation=True, by_probs=False,
            label="Ensembled accuracy")
    else:
        print(f"Ensemble file {ensemble_filename} does not exist.")
```
