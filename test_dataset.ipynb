```python
from dataset import MyriadLama100Dataset


dataset = MyriadLama100Dataset(model_name="llama3.2_3b", debug=True)
```

```python
dataloader = dataset.get_dataloader(batch_size=8, shuffle=False)
```

```python
import random

num_samples = 5
num_paraphrases = 3

all_samples = []
for uuids, answers, paraphrases, _ in dataloader:
    paraphrases = list(zip(*paraphrases))
    for uuid, answers, paraphrases_ in zip(uuids, answers, paraphrases):
        random.seed(uuid)
        paraphrases_ = list(paraphrases_)
        random.shuffle(paraphrases_)
        assert num_samples * num_paraphrases <= len(paraphrases_)

        para_groups = [paraphrases_[i * num_paraphrases:(i + 1) * num_paraphrases] for i in range(num_samples)]
        for para_group in para_groups:
            all_samples.append((uuid, answers, para_group))

```
